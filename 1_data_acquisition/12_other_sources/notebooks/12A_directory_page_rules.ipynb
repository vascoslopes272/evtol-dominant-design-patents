# 12A. Reading the evtol.news pages and the AAM Reality Index

For every aircraft of the evtol.news directory this notebook applies the keyword rules of `scripts/yearscript/` to the text of its public page and writes four lists: whether the aircraft is inside the domain (`DOMAIN`), the year it first appears (`YEAR`), its technology readiness level (`TRL`) and the company of the AAM Reality Index that matches each assignee of the patent dataset (`COMPANY_MAP`).

- **Reads:** the evtol.news pages and sentences in `eVTOL_dataset/06_external/61_evtol_news_directory`, the AAM index in `62_aam_index`, and the design table of the dataset.
- **Writes:** `2_analysis/review/script_outputs/{DOMAIN,YEAR,TRL,COMPANY_MAP}.csv`. The analysis reads them, with the author's own reading where the author reviewed an item (`2_analysis/review/decisions`).
- **Rules:** `year_rule.py` picks the earliest year of an event of the aircraft from the sentences that contain a year, `trl_rule.py` and `trl_page_rule.py` read the readiness level, `ari_rule.py` matches the company names. No model is used.
- The accuracy of each rule on the author's review sample is in `eVTOL_dataset/06_external/61_evtol_news_directory/directory_accuracy.csv`.
