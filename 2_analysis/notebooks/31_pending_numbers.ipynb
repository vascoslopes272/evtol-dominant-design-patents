# 31. Pending numbers of paper P1

Runs, in order, the scripts that produce the numbers the paper asks for beyond notebooks 21-29: rarefaction, variety threshold and share window (39, 43, 38), levels and the last 100 designs (41), family F (31, 42), regions, tables S1, S2 and the Spearman list (45-48), locks, query funnel, samples and the Gower example (55), the coding-noise and weight results (36, 40) and the assembly (44). Output: `results/2_analysis/pending_numbers/pending_numbers.md` and `pending_csv/`. The status of every request is in `docs/request_trace.md`.

The coding-noise numbers (scripts 36 and 40) read the replica files of `3_robustness/31_label_reliability` (`allowed_r15_p10000`, `allowed_r15_p1000_n1000`, `allowed_r20_p10000`); run `31B_run_replicas.py` first (about 5 h for 120 replicas at 10 000 permutations).

**Depends on:** 21-29 (freeze tables), `3_robustness/31_label_reliability`, `3_robustness/32_weight_sensitivity` results.

```python
import os, subprocess, sys, pathlib
REPO = pathlib.Path.cwd().resolve().parents[1]
env = dict(os.environ, PYTHONPATH=str(REPO / '2_analysis' / 'src'))
SCRIPTS = ['31_family_F_numbers.py', '38_parameter_sweep.py', '38_parameter_sweep.py', '38_parameter_sweep.py', '38_parameter_sweep.py', '39_rarefaction_verdict_table.py', '41_levels_and_last100.py', '42_family_F_trends.py', '43_variety_window_tables.py',
           '45_pending_D_items_11_12.py', '46_pending_D_item_13_table_S1.py', '47_pending_D_item_14_table_S2.py', '48_pending_D_item_20_spearman_holm.py',
           '55_pending_part_D2.py', '36_robustness_numbers.py', '40_pending_block_A.py', '44_assemble_pending_numbers.py']
ARGS = [None, ['--part', 'rare', '--n', '30'], ['--part', 'rare', '--n', '50'], ['--part', 'variety'], ['--part', 'window']]
for k, s in enumerate(SCRIPTS):
    r = subprocess.run([sys.executable, str(REPO / '2_analysis' / 'scripts' / s), *(ARGS[k] or [] if k < len(ARGS) else [])], env=env, capture_output=True, text=True)
    print(s, 'exit', r.returncode, (r.stderr or r.stdout)[-300:].replace('\n', ' '))
```
