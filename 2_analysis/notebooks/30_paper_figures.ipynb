# 30. Paper figures

Every panel of the paper figures (4 to 7) as a separate vector PDF of 85 x 65 mm, plus the three balance panels together (`fig5abc_balance_three_levels.pdf`, 183 x 65 mm) and the disparity panel in the thesis reading (`fig5d_disparity_thesis_reading.pdf`). The code is `scripts/33_paper_figures.py`; it reads the tables of notebooks 21-29 and of `scripts/31_family_F_numbers.py` (run that first; its permutation count is written in `results/2_analysis/paper_figures/log.txt`).

**Depends on:** 22, 23, 27 (tables), `scripts/31_family_F_numbers.py`.

```
import os, subprocess, sys, pathlib
REPO = pathlib.Path.cwd().resolve().parents[1]
env = dict(os.environ, PYTHONPATH=str(REPO / '2_analysis' / 'src'))
r = subprocess.run([sys.executable, str(REPO / '2_analysis' / 'scripts' / '33_paper_figures.py')], env=env, capture_output=True, text=True)
print(r.stdout[-3000:], r.stderr[-1500:])
```

```
OUT = REPO / 'results' / '2_analysis' / 'paper_figures'
for f in sorted(OUT.glob('*.pdf')):
    print(f'{f.stat().st_size / 1024:6.0f} kB  {f.name}')
```
